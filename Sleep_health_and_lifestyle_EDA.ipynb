### **Sleep Health and Lifestyle – Exploratory Data Analysis (EDA)**

- This project focuses on analyzing the relationship between sleep patterns, lifestyle habits, and health-related factors using the Sleep Health and Lifestyle Dataset. The main objective of this analysis is to understand how different lifestyle parameters such as age, gender, occupation, physical activity level, stress level, BMI category, heart rate, daily steps, and sleep disorders influence sleep quality and overall health.

- The project follows the complete EDA workflow, including data loading, data exploration, data cleaning, statistical analysis, and data visualization. Python libraries such as NumPy, Pandas, Matplotlib, Seaborn, and Statistics were used for data processing and visualization.

- Various analytical techniques were applied to understand the dataset, including distribution analysis, frequency analysis, correlation analysis, and graphical representation. Different visualizations such as bar charts, pie charts, histograms, box plots, scatter plots, and density plots were created to identify important patterns and trends in sleep behavior.

- The analysis helps identify relationships between physical activity, daily steps, age, sleep duration, and other health indicators. This project provides valuable insights into how lifestyle choices affect sleep health and can help in making better health-related decisions.

### **Problem Statement**
- In today's fast-paced environment, poor sleep quality and sleep disorders are becoming increasingly common, significantly impacting overall health and productivity. The objective of this project is to analyze the Sleep Health and Lifestyle dataset to uncover the relationships between an individual's daily habits, occupation, and physiological metrics with their overall sleep health. By exploring this data, we aim to identify the key lifestyle factors—such as physical activity, stress levels, and BMI—that most strongly influence sleep duration, sleep quality, and the prevalence of sleep disorders.

#### **Key Objectives of the Analysis**
- To solve this problem and guide your EDA, your project can aim to answer the following core questions:

   - Occupational Impact: How do different professions affect a person's stress levels and average sleep duration?

   - Lifestyle Correlations: Is there a direct, measurable relationship between daily physical activity (like daily steps) and improved sleep quality?

   - Health Risk Factors: How do physiological markers like BMI Category, Blood Pressure, and Heart Rate correlate with the likelihood of having a Sleep Disorder (like Sleep Apnea or Insomnia)?

  - Demographic Differences: Are there significant variations in sleep patterns, stress levels, and sleep disorders when comparing genders and age groups?

### **Daily Steps Distribution**

```python
import numpy as np
```

```python
plt.figure(figsize=(13, 6))
counts, bin_edges = np.histogram(shls_df["Daily Steps"], bins=10)
bin_midpoints = (bin_edges[:-1] + bin_edges[1:]) / 2
plt.plot(bin_midpoints, counts, color='steelblue', marker='o', linewidth=2, label='Daily Steps')

for i in range(len(bin_midpoints)):
    plt.text(bin_midpoints[i], counts[i] + 2, str(counts[i]), 
             ha='center', va='bottom', fontsize=15, color='black')
    
plt.title("Daily Steps Distribution",fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Daily Steps",fontsize=16)
plt.ylabel("Frequency",fontsize=16)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

### **insight: Daily Steps Distribution (Line Histogram)**
- This graph shows the distribution of daily physical activity among participants.
- The X-axis represents the range of daily steps, while the Y-axis represents the number of participants (frequency) in each range.
- The peak point indicates the step range where most participants have similar daily activity levels.
- If the graph is concentrated in a particular range, it suggests that most participants follow a similar activity pattern.
- Very low or very high daily step values may indicate less active or highly active individuals and can be further checked for outliers.
- Daily steps can help analyze the relationship between physical activity, sleep quality, stress level, and overall lifestyle health.

### **Sleep Duration**

```python
plt.figure(figsize=(6,4))
plt.boxplot(shls_df["Sleep Duration"])
plt.title("Box Plot of Sleep Duration",fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Sleep Duration")
plt.ylabel("Hours")
plt.show()
```

### **Relationship Between Age and Sleep Duration**

```python
plt.figure(figsize=(8,5))
sns.regplot(data=shls_df,x="Age",y="Sleep Duration",scatter_kws={"color":"teal","s":60},line_kws={"color":"y","linewidth":2})
plt.title("Relationship Between Age and Sleep Duration", fontsize=14)
plt.xlabel("Age (Years)")
plt.ylabel("Sleep Duration (Hours)")
plt.show()
plt.savefig("Regression Line")
```

### **insight: Relationship Between Age and Sleep Duration**
- There is a weak positive relationship between Age and Sleep Duration.
- Most individuals sleep between 6 and 8 hours, regardless of age.
- The regression line shows that sleep duration changes only slightly with increasing age.
- The wide spread of data points indicates that age alone is not a strong predictor of sleep duration.
- Other factors such as stress level, physical activity, occupation, and sleep disorders may have a greater influence on sleep duration.

### **Physical Activity vs Daily Steps Graph**

```python
plt.figure(figsize=(8,5))
sns.regplot(data=shls_df,x="Physical Activity Level",y="Daily Steps",)
plt.title("Physical Activity Level vs Daily Steps")
plt.xlabel("Physical Activity Level")
plt.ylabel("Daily Steps")
plt.show()
```

- **Insight**
* A positive relationship exists between physical activity level and daily steps.
* Participants with higher activity levels generally record more daily steps.

### **Age vs Sleep Duration by Gender**

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=shls_df,x="Age",y="Sleep Duration",hue="Gender",palette={0:"green",1:"red"},s=80)
plt.title("Age vs Sleep Duration by Gender")
plt.xlabel("Age (Years)")
plt.ylabel("Sleep Duration (Hours)")
plt.legend(labels=["Female","Male"])
plt.show()
plt.savefig("Scatterplot__Age_vs_Sleep_Duration_by_Gender")
```

### **insight**
- This graph compares sleep duration across different ages for males and females.
- Green points represent females and red points represent males.
- It helps identify whether one gender tends to sleep more or less at similar ages.
